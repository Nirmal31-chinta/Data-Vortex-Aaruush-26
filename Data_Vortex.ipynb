```python
# Create copies so the original datasets remain unchanged
posts_clean = posts.copy()
users_clean = users.copy()

# Remove exact duplicate rows from Posts
before = len(posts_clean)

posts_clean = posts_clean.drop_duplicates()

after = len(posts_clean)

print("Posts before removing duplicates:", before)
print("Posts after removing duplicates:", after)
print("Duplicate rows removed:", before - after)
```

```python
# Check NULL markers after duplicate removal

null_text = posts_clean[
    posts_clean["text_content"].astype(str).str.strip().str.contains(
        r"^NULL",
        case=False,
        regex=True,
        na=False
    )
]

print("NULL-type text values:", len(null_text))

print("\nExamples:")
display(null_text[["post_id", "text_content"]].head(10))
```

```python
# Convert corrupted NULL-type text values into actual missing values

null_mask = posts_clean["text_content"].astype(str).str.strip().str.contains(
    r"^NULL",
    case=False,
    regex=True,
    na=False
)

posts_clean.loc[null_mask, "text_content"] = pd.NA

print("NULL-type values converted to missing:", null_mask.sum())

print("\nMissing text values after this step:")
print(posts_clean["text_content"].isna().sum())
```

```python
import re
import html

def clean_text(text):
    if pd.isna(text):
        return text

    text = str(text)

    # Remove HTML tags such as <div>, <br>, etc.
    text = re.sub(r"<[^>]+>", "", text)

    # Decode HTML entities such as &amp;, &quot;, etc.
    text = html.unescape(text)

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text

# Apply the cleaning function
posts_clean["text_content"] = posts_clean["text_content"].apply(clean_text)

print("Text cleaning applied successfully.")

print("\nExamples after cleaning:")
display(posts_clean[["post_id", "text_content"]].head(10))
```

```python
# Check for remaining HTML tags or HTML entities

remaining_html = posts_clean[
    posts_clean["text_content"].astype(str).str.contains(
        r"<[^>]+>|&(?:amp|lt|gt|quot|apos);|&#\d+;",
        regex=True,
        na=False
    )
]

print("Rows with remaining HTML/encoded artifacts:", len(remaining_html))

if len(remaining_html) > 0:
    print("\nExamples:")
    display(remaining_html[["post_id", "text_content"]].head(20))
```

```python
# Check remaining HTML/encoded artifacts, ignoring missing values

remaining_html = posts_clean[
    posts_clean["text_content"].notna()
    & posts_clean["text_content"].str.contains(
        r"<[^>]+>|&(?:amp|lt|gt|quot|apos);|&#\d+;",
        regex=True,
        na=False
    )
]

print("Rows with remaining HTML/encoded artifacts:", len(remaining_html))

if len(remaining_html) > 0:
    print("\nExamples:")
    display(remaining_html[["post_id", "text_content"]].head(20))
```

```python
def convert_timestamp(value):
    if pd.isna(value):
        return pd.NaT

    value = str(value).strip()

    # 10-digit Unix timestamp
    if re.fullmatch(r"\d{10}", value):
        return pd.to_datetime(int(value), unit="s", errors="coerce")

    # DD-MM-YYYY format
    if re.fullmatch(r"\d{2}-\d{2}-\d{4}", value):
        return pd.to_datetime(
            value,
            format="%d-%m-%Y",
            errors="coerce"
        )

    # ISO format or other standard datetime representation
    return pd.to_datetime(value, errors="coerce")


# Convert timestamps
posts_clean["timestamp"] = posts_clean["timestamp"].apply(convert_timestamp)

print("Missing/invalid timestamps after conversion:")
print(posts_clean["timestamp"].isna().sum())

print("\nTimestamp examples after conversion:")
print(posts_clean["timestamp"].head(10))
```

```python
# Check for suspicious text-encoding artifacts

encoding_pattern = r"Ã|Â|â€™|â€œ|â€|ð"

suspicious_text = posts_clean[
    posts_clean["text_content"].notna()
    & posts_clean["text_content"].astype(str).str.contains(
        encoding_pattern,
        regex=True,
        na=False
    )
]

print("Number of posts with suspicious encoding artifacts:")
print(len(suspicious_text))

print("\nExamples:")
print(suspicious_text["text_content"].head(10).tolist())
```

```python
# Test UTF-8 encoding repair on suspicious text

def fix_encoding(text):
    if pd.isna(text):
        return text

    text = str(text)

    try:
        return text.encode("latin1").decode("utf-8")
    except (UnicodeEncodeError, UnicodeDecodeError):
        return text


# Apply only to suspicious rows for testing
encoding_test = suspicious_text["text_content"].apply(fix_encoding)

print("Before → After:\n")

for before, after in zip(
    suspicious_text["text_content"].head(10),
    encoding_test.head(10)
):
    print("BEFORE:", before)
    print("AFTER :", after)
    print()
```

```python
# Repair suspicious text-encoding artifacts

encoding_mask = (
    posts_clean["text_content"].notna()
    & posts_clean["text_content"].astype(str).str.contains(
        encoding_pattern,
        regex=True,
        na=False
    )
)

# Count affected rows before repair
affected_before = encoding_mask.sum()

# Apply the repair
posts_clean.loc[encoding_mask, "text_content"] = (
    posts_clean.loc[encoding_mask, "text_content"].apply(fix_encoding)
)

print("Posts repaired:", affected_before)
```

```python
# Verify that suspicious encoding artifacts are gone

remaining_encoding = posts_clean[
    posts_clean["text_content"].notna()
    & posts_clean["text_content"].astype(str).str.contains(
        encoding_pattern,
        regex=True,
        na=False
    )
]

print("Remaining suspicious encoding artifacts:", len(remaining_encoding))

if len(remaining_encoding) > 0:
    print("\nRemaining examples:")
    print(remaining_encoding["text_content"].head(10).tolist())
else:
    print("Text encoding check passed: no suspicious artifacts remain.")
```
